## What is TF-IDF Vectorizer?
TF-IDF Vectorizer is an NLP technique that converts text documents into numerical feature vectors by assigning weights to words based on their frequency in a document and their rarity across the document collection.

```python
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
pd.set_option('display.max_colwidth', None)
```

```python
corpus = new_df['tags']
```

```python
cv = CountVectorizer(max_features=5000, stop_words='english')
cv_vectors = cv.fit_transform(corpus)  
```

```python
tfidf = TfidfVectorizer(max_features=5000, stop_words='english', sublinear_tf=True)
tfidf_vectors = tfidf.fit_transform(corpus)              
```

```python
print("CountVectorizer shape:", cv_vectors.shape)
print("TF-IDF shape:        ", tfidf_vectors.shape)
```

```python
title_to_idx = pd.Series(np.arange(len(new_df)), index=new_df['title'].values)
title_to_idx = title_to_idx[~title_to_idx.index.duplicated()]
```

```python
def top_k(vectors, idx, k=5):
    scores = cosine_similarity(vectors[idx], vectors).ravel()
    scores[idx] = -1                                     
    top = np.argpartition(-scores, k)[:k]
    return top[np.argsort(-scores[top])]
```

```python
genre_map = dict(zip(movies['movie_id'], movies['genres']))
def genre_jaccard(i, j):
    gi = set(genre_map[new_df.at[i, 'movie_id']])
    gj = set(genre_map[new_df.at[j, 'movie_id']])
    return len(gi & gj) / max(len(gi | gj), 1)
```

```python
# A sparse matrix is a matrix where the majority of elements are zero, 
# typically containing only a small fraction of non-zero values.
```

```python
test_movies = ['Avatar', 'The Dark Knight Rises', 'Silent Running', 'Spectre', 'Toy Story']
```

```python
rows = []
for title in test_movies:
    if title not in title_to_idx.index:
        print(f"Skip: '{title}' not present in dataset ")
        continue
    idx = title_to_idx[title]
    for name, vecs in [('CountVectorizer', cv_vectors), ('TF-IDF', tfidf_vectors)]:
        top = top_k(vecs, idx, k=5)
        overlap = np.mean([genre_jaccard(idx, j) for j in top])
        rows.append({
            'movie': title,
            'model': name,
            'avg_genre_overlap': round(overlap, 3),
            'top5': ', '.join(new_df['title'].iloc[top].tolist()),
        })
result = pd.DataFrame(rows)
```

```python
print("\n Movie-wise results: ")
print(result.to_string(index=False))
```

```python
print("\n Average genre overlap: ")
print(result.groupby('model')['avg_genre_overlap'].mean().round(3))
```

```python
print((new_df.index == range(len(new_df))).all())   
```

```python
best_name, best_vectors = 'TF-IDF', tfidf_vectors   
```

```python
best_name = (
    result.groupby('model')['avg_genre_overlap']
    .mean()
    .idxmax()
)

best_vectors = {
    'TF-IDF': tfidf_vectors,
    'CountVectorizer': cv_vectors
}[best_name]

print("Best model:", best_name)
```

```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

K = 20
n = len(new_df)
neighbors = np.zeros((n, K), dtype=np.int32)

for start in range(0, n, 500):             
    end = min(start + 500, n)
    sims = cosine_similarity(best_vectors[start:end], best_vectors)
    for r, i in enumerate(range(start, end)):
        sims[r, i] = -1                    
        top = np.argpartition(-sims[r], K)[:K]
        neighbors[i] = top[np.argsort(-sims[r][top])]
```

```python
import pickle
import os 

artifacts = {
    'movies': new_df[['movie_id', 'title']].reset_index(drop=True),
    'neighbors': neighbors,
    'vectorizer': tfidf,       # future mein naya text/query transform karne ke liye
    'model_name': best_name,
}

with open('movie_recommender.pkl', 'wb') as f:
    pickle.dump(artifacts, f, protocol=pickle.HIGHEST_PROTOCOL)

print("Saved. Size (MB):", round(os.path.getsize('movie_recommender.pkl') / 1e6, 2))
```

```python
with open('movie_recommender.pkl', 'rb') as f:
    art = pickle.load(f)

movies_tbl = art['movies']
nbrs = art['neighbors']
title_map = pd.Series(np.arange(len(movies_tbl)), index=movies_tbl['title'].values)
title_map = title_map[~title_map.index.duplicated()]

def recommend(title, k=5):
    idx = title_map[title]
    return movies_tbl['title'].iloc[nbrs[idx][:k]].tolist()

print(recommend('Avatar'))
print(recommend('The Dark Knight Rises'))

# Consistency check: reloaded neighbours == fresh computation
i = title_map['Avatar']
fresh = top_k(best_vectors, i, k=5)
print("Match:", np.array_equal(nbrs[i][:5], fresh))
```
